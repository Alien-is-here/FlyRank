## 3. The flag-linked test

*Pick a signal one of FlyRank's real flags relies on. Does the data support the rule's assumption?*

**Answer:**

The audit shows that no single signal explains page performance by itself. Older, high-traffic pages that are declining are reasonable candidates for review, but the signals should be treated as decision-support rather than proof that refreshing a page will improve traffic. The content team should review the highest-ranked pages first and consider other factors before making a refresh decision.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
```

```python
print("=== FLAG-LINKED TEST ===")

down_pages = df[df["trend_direction"] == "down"]

print(f"Declining pages: {len(down_pages):,}")
print(f"Average impressions of declining pages: {down_pages['impressions_90d'].mean():.2f}")
print(f"Median impressions of declining pages: {down_pages['impressions_90d'].median():.2f}")

high_volume_down = down_pages[
    down_pages["impressions_90d"] >= 1000
]

print(f"High-volume + declining pages: {len(high_volume_down):,}")
```

## 4. What this means in practice

*Two or three sentences: what a content team should take from this.*

**Answer:**

The content team should prioritize pages that are old, have meaningful traffic, and are showing a decline. These signals help identify pages worth reviewing, but the score is only a guide and does not guarantee that a refresh will improve performance.
