## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

**Answer:**


ML could potentially rank pages better than a simple fixed rule if it can learn useful relationships that the rule misses. This should be tested by comparing the ML ranking against the baseline.


A fixed rule would require us to manually decide which characteristics define each type of content. However, content performance can depend on several factors at the same time, and the patterns may not be simple enough for a few if-statements.

A **simple rule** would be:
```python
if age >= 90 and impressions >= 1000 and trend == 'down':
    score = age * impressions
```

**But this is too rigid:**
- A page at 89 days old vs 90 days old gets treated very differently (hard cutoff is dumb)
- Maybe CTR matters too — a page with low CTR needs different treatment than high CTR
- Maybe very very old pages (300+ days) aren't worth refreshing at all

**ML can learn:**
- Soft boundaries instead of hard cutoffs
- How age interacts with traffic and CTR
- Special cases (like when age doesn't matter)

So ML should rank pages better than a simple rule.
